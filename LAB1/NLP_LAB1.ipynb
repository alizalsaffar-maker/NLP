## Task #2 — Download and open the dataset

Below we:
1. Import the libraries we need.
2. Download the file from the internet and save a local copy (so it is a real, reusable dataset file on disk, not just an in-memory read).
3. Open/load it using `pandas`.
4. Display the first few rows to prove it opened correctly.
5. Do a little extra exploration (shape, columns, label counts) to confirm the data makes sense.

```python
# Step 1: Import the libraries we need
# - pandas: the standard Python library for loading and working with tabular data (CSV/TSV/JSON/SQL, etc.)
# - urllib.request: a built-in Python library used to download files from a URL
import pandas as pd
import urllib.request
import os
```

```python
# Step 2: Download the dataset from the internet

# The 'raw' GitHub link gives us the plain text file, not the GitHub webpage
dataset_url = "https://raw.githubusercontent.com/sharmaroshan/Restaurant-Reviews-Analysis/master/Restaurant_Reviews.tsv"

# This is the name we will save the file under, in our current working folder
local_filename = "Restaurant_Reviews.tsv"

# urlretrieve() downloads whatever is at 'dataset_url' and saves it to 'local_filename'
urllib.request.urlretrieve(dataset_url, local_filename)

# Confirm the file was actually downloaded and saved to disk
if os.path.exists(local_filename):
    file_size_kb = os.path.getsize(local_filename) / 1024
    print(f"Dataset downloaded successfully: '{local_filename}' ({file_size_kb:.1f} KB)")
else:
    print("Download failed — please check your internet connection.")
```

```python
# Step 3: Open the dataset as raw text first, just to prove it is a plain, human-readable text file
# (This satisfies the "Text Files" idea from the lab: any dataset can first be viewed as raw lines of text.)
with open(local_filename, "r", encoding="utf-8") as f:
    first_lines = [next(f) for _ in range(5)]

print("First 5 raw lines of the file:\n")
for line in first_lines:
    print(line.strip())
```

```python
# Step 4: Properly load the dataset into a pandas DataFrame
# sep='\t' tells pandas this file is TAB-separated (a .tsv) rather than comma-separated (.csv)
df = pd.read_csv(local_filename, sep="\t")

# Basic checks to confirm the dataset opened correctly
print("Dataset shape (rows, columns):", df.shape)
print("Column names:", list(df.columns))
```

```python
# Step 5: Display the first few rows of the dataset to prove it was successfully opened
df.head()
```

```python
# Step 6 (extra exploration): quick summary of the labels
# 'Liked' = 1 means a positive review, 'Liked' = 0 means a negative review
print("Label distribution (0 = negative, 1 = positive):")
print(df["Liked"].value_counts())

# Show one example positive review and one example negative review
print("\nExample positive review:")
print(df[df["Liked"] == 1]["Review"].iloc[0])

print("\nExample negative review:")
print(df[df["Liked"] == 0]["Review"].iloc[0])
```
