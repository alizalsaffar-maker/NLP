# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


**Answer:** A decoder-only model is well suited to a chatbot because it is trained to do exactly what a conversation requires: predict the next token given everything said so far. Generating a reply is just continuing that sequence one token at a time, using the whole conversation history as context, so no separate step is needed to first "encode" the input into a fixed representation the way an encoder-decoder model does. This autoregressive generation naturally handles open-ended, variable-length output -- a chatbot reply can be a single word or several paragraphs, and the model doesn't need to know in advance how long or what shape the "correct" answer is, unlike a fixed task such as translation or classification. It also scales naturally to multi-turn dialogue: each new message is simply appended to the growing sequence the model conditions on. This flexibility is exactly why GPT-style decoder-only architectures power conversational systems like ChatGPT.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
# Optional Challenge demonstration: sampling with vs. without a fixed seed
from transformers import set_seed

prompt = "Natural language processing is"

print("-- With set_seed(42) before every call: output should be identical each time --")
for i in range(2):
    set_seed(42)
    out = generator(prompt, max_new_tokens=30, do_sample=True)
    print(f"Run {i+1}:", out[0]["generated_text"])

print("\n-- Without set_seed: outputs can differ each time --")
for i in range(2):
    out = generator(prompt, max_new_tokens=30, do_sample=True)
    print(f"Run {i+1}:", out[0]["generated_text"])
````

**Answer:** With `do_sample=True`, the model doesn't always pick the single most likely next token (as `do_sample=False`/greedy decoding does) -- instead, at each step it samples randomly from the probability distribution over possible next tokens, so a less-likely word can still get chosen by chance. `set_seed(42)` fixes Python's/PyTorch's random number generator to a specific starting point, so that "random" choice becomes reproducible and every run produces the same output. Without calling `set_seed`, each run draws from a different random sequence, so the sampled tokens -- and therefore the generated text -- can differ from one run to the next even with the exact same prompt.
